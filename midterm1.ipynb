```python
import pandas as pd
from IPython.display import Markdown, display

display(Markdown(open("my_data/DATA.md", encoding="utf-8").read()))
ranks = pd.read_csv("my_data/retrieval_ranks.csv")   # Part 1
docs = pd.read_csv("my_data/doc_results.csv")        # Part 2
display(ranks.head(), docs)
```

```python
import math

def wilson(hits, n, z=1.96):
    p = hits / n
    d = 1 + z**2 / n
    centre = (p + z**2 / (2*n)) / d
    half = z * math.sqrt(p*(1-p)/n + z**2/(4*n**2)) / d
    return centre - half, centre + half
```

```python
# 1a: recall@1, @3, @5 and MRR@10 for each config
for cfg in ["A", "B"]:
    r = ranks["rank_" + cfg]
    print(cfg,
          "recall@1", (r <= 1).mean(),
          "recall@3", (r <= 3).mean(),
          "recall@5", (r <= 5).mean(),
          "MRR@10", (1 / r.where(r <= 10)).fillna(0).mean())
```

```python
# 1b: recall and Wilson interval of each candidate
for cfg, k in [("A", 3), ("A", 5), ("B", 3), ("B", 5)]:
    hits = int((ranks["rank_" + cfg] <= k).sum())
    lo, hi = wilson(hits, 60)
    print(cfg, k, hits / 60, lo, hi)
```

```python
langs = ["EN", "RU", "KK"]
share = {"EN": 0.20, "RU": 0.41, "KK": 0.39}
q = {"EN": 24, "RU": 37, "KK": 78}
a = {"EN": 146, "RU": 223, "KK": 473}
chunk = {"A": {"EN": 248, "RU": 379, "KK": 804},
         "B": {"EN": 70, "RU": 107, "KK": 227}}
system = 318
p_in, p_out = 3, 18
req_per_day = 9000
budget = 32900

for cfg, k in [("A", 3), ("A", 5), ("B", 3), ("B", 5)]:
    per_request = 0
    for lg in langs:
        tokens_in = system + q[lg] + k * chunk[cfg][lg]
        cost = (tokens_in * p_in + a[lg] * p_out) / 1e6
        per_request += share[lg] * cost
    yearly = per_request * req_per_day * 365
    print(cfg, k, round(yearly), "OK" if yearly <= budget else "over budget")
```

```python
for cfg, k in [("A", 3), ("A", 5), ("B", 3), ("B", 5)]:
    for lang in ["EN", "RU", "KK"]:
        rows = ranks[ranks["lang"] == lang]
        hits = (rows["rank_" + cfg] <= k).sum()
        n = len(rows)
        low, high = wilson(hits, n)
        print(cfg, k, lang, hits, n, round(hits / n, 3), round(low, 3), round(high, 3))
```

```python
import math

for pipe in ["H", "L"]:
    d = docs[docs["pipeline"] == pipe]
    hits = d["n_correct"].sum()
    n = d["n_docs"].sum()
    low_all, high_all = wilson(hits, n)
    print(pipe, "overall", hits, n, round(hits / n, 3), round(low_all, 3), round(high_all, 3))
    worst = d.loc[(d["n_correct"] / d["n_docs"]).idxmin()]
    low, high = wilson(worst["n_correct"], worst["n_docs"])
    print(pipe, "weakest", worst["lang"], worst["condition"], worst["n_correct"],
          worst["n_docs"], round(worst["n_correct"] / worst["n_docs"], 3),
          round(low, 3), round(high, 3))

pages_per_day = 5500
tokens = math.ceil(827 / 28) * math.ceil(1169 / 28)
cost_h = pages_per_day * 365 * (tokens * 3 + 318 * 15) / 1e6
servers = math.ceil(pages_per_day / 8000)
cost_l = servers * 450 * 12
pii = pages_per_day * 365 * 0.27
print("tokens per page:", tokens)
print("H annual cost:", round(cost_h))
print("L servers and annual cost:", servers, cost_l)
print("PII pages per year:", round(pii))
```
